```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import seaborn as sns
import matplotlib.pyplot as plt


#For pretty display
pd.set_option('display.max_columns', None)

```

```python
file_path = r"C:\My stuff\Coding\ML project\KiranveerSingh_Project\Dataset\nutrition_estimator_factors.xlsx"
df = pd.read_excel(file_path, engine='openpyxl')
# Define thresholds for deficiency (adjust as needed)
thresholds = {
    'Daily_Calorie_Intake': 2000,
    'Daily_Protein_Intake_(g)': 50,
    'Daily_Fat_Intake_(g)': 70,
    'Daily_Carbohydrate_Intake_(g)': 300,
    'Daily_Fiber_Intake_(g)': 25,
    'Daily_Water_Intake_(liters)': 2.7,
    'Vitamin_A_Intake_(mcg)': 900,
    'Vitamin_B1_(Thiamine)_Intake_(mg)': 1.2,
    'Vitamin_B2_(Riboflavin)_Intake_(mg)': 1.3,
    'Vitamin_B3_(Niacin)_Intake_(mg)': 16,
    'Vitamin_B5_(Pantothenic_Acid)_Intake_(mg)': 5,
    'Vitamin_B6_Intake_(mg)': 1.3,
    'Vitamin_B12_Intake_(mcg)': 2.4,
    'Vitamin_C_Intake_(mg)': 90,
    'Vitamin_D_Intake_(IU)': 600,
    'Vitamin_E_Intake_(mg)': 15,
    'Vitamin_K_Intake_(mcg)': 120,
    'Iron_Intake_(mg)': 8,
    'Zinc_Intake_(mg)': 11,
    'Magnesium_Intake_(mg)': 400,
    'Calcium_Intake_(mg)': 1000,
    'Iodine_Intake_(mcg)': 150,
    'Selenium_Intake_(mcg)': 55,
    'Copper_Intake_(mg)': 0.9,
    'Folate_(Folic_Acid)_Intake_(mcg)': 400,
    'Phosphorus_Intake_(mg)': 700,
    'Manganese_Intake_(mg)': 2.3,
    'Chromium_Intake_(mcg)': 35,
    'Food_Diversity_Score': 5,
    'Hemoglobin_(g/dL)': 13.5,
    'Serum_Ferritin_(ng/mL)': 30,
    'Vitamin_D_25-OH_(ng/mL)': 30,
    'Serum_B12_(pg/mL)': 200,
    'Serum_Folate_(ng/mL)': 3,
    'C-Reactive_Protein_(mg/dL)': 3,  # high is bad, but here low is not a deficiency
    'Serum_Albumin_(g/dL)': 3.5,
    'Total_Cholesterol_(mg/dL)': 125,  # let's use 125 as minimum, lower may be bad
    'Fasting_Glucose_(mg/dL)': 70,  # fasting glucose below 70 may indicate hypoglycemia
}

# Create a deficiency column that lists deficient features
def find_deficiencies(row):
    deficiencies = []
    for feature, threshold in thresholds.items():
        if feature in row and pd.notna(row[feature]):
            if row[feature] < threshold:
                deficiencies.append(feature)
    return deficiencies

df['Deficiency_Features'] = df.apply(find_deficiencies, axis=1)

# Optional: Also add count of deficiencies
df['Deficiency_Count'] = df['Deficiency_Features'].apply(len)

# Save to Excel
df.to_excel(r"C:\My stuff\Coding\ML project\KiranveerSingh_Project\Dataset\nutrition_data_with-deficiencies.xlsx", index=False)
print("Saved with named deficiency features.")
```
