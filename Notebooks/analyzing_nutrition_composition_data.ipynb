```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load cleaned food nutrition dataset
df = pd.read_csv(r"C:\My stuff\Coding\ML project\KiranveerSingh_Project\Dataset\cleaned_food_nutrition_dataset.csv")
df.head()
```

```python
#Distribution Plot for Calories
sns.histplot(df['Calories'], kde=True, bins=30, color='skyblue')
plt.title("Distribution of Calories")
plt.xlabel("Calories")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
#Correlation Heatmap
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap of Nutrients")
plt.tight_layout()
plt.show()
```

```python
#Bar Plot – Top 20 Foods by Protein
top_protein = df.sort_values(by='Protein_g', ascending=False).head(20)
sns.barplot(data=top_protein, y='Food_Item', x='Protein_g', palette='viridis')
plt.title("Top 20 Foods by Protein Content")
plt.xlabel("Protein (g)")
plt.ylabel("Food Item")
plt.tight_layout()
plt.show()
```

```python
#Pie Chart – Food Category Distribution
category_counts = df['Category'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(category_counts, labels=category_counts.index, autopct='%1.1f%%', startangle=140)
plt.title("Food Category Distribution")
plt.axis('equal')
plt.tight_layout()
plt.show()
```

```python
#Violin Plot – Distribution of Calcium
sns.violinplot(data=df, y='Calcium_mg', color='lightgreen')
plt.title("Violin Plot of Calcium")
plt.tight_layout()
plt.show()
```

```python
#Box Plot – Iron Across Categories
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x='Category', y='Iron_mg')
plt.xticks(rotation=45)
plt.title("Box Plot of Iron Content by Category")
plt.tight_layout()
plt.show()
```

```python
#Scatter Plot – Calories vs Protein
sns.scatterplot(data=df, x='Calories', y='Protein_g', hue='Category')
plt.title("Calories vs Protein")
plt.tight_layout()
plt.show()
```

```python
#Count Plot – Categories
sns.countplot(data=df, y='Category', order=df['Category'].value_counts().index, palette='Set2')
plt.title("Count of Food Items by Category")
plt.tight_layout()
plt.show()
```

```python
#KDE Plot – Fat Distribution
sns.kdeplot(df['Fat_g'], shade=True, color='orange')
plt.title("KDE Plot of Fat")
plt.xlabel("Fat (g)")
plt.tight_layout()
plt.show()
```
