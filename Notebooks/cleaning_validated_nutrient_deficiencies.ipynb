```python
import pandas as pd

# Load the validated deficiencies file
file_path = r"C:\My stuff\Coding\ML project\KiranveerSingh_Project\Dataset\validated_nutrient_deficiencies.csv"
df = pd.read_csv(file_path)


# Remove completely duplicate rows
df = df.drop_duplicates().reset_index(drop=True)

# Standardize Gender (capitalize, e.g. Male/Female)
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].astype(str).str.strip().str.capitalize()

# Standardize and clean key categorical columns if present
for col in [
    'Allergies_or_Intolerances',
    'Food_Preferences_(e.g._Vegan,_Gluten-Free)'
]:
    if col in df.columns:
        df[col] = df[col].astype(str).str.lower().str.strip().replace({'nan': '', 'none': ''})

# Strip string columns of leading/trailing whitespace
for col in df.select_dtypes('object'):
    df[col] = df[col].str.strip()

# Identify numeric columns (nutrient intakes, lab values)
numeric_cols = [
    col for col in df.columns
    if (col.endswith('Intake_(mg)')
        or col.endswith('Intake_(g)')
        or col.endswith('Intake_(mcg)')
        or col.endswith('Deficiency')
        or col.endswith('Score')
        or col in [
            'Height_(cm)', 'Weight_(kg)', 'BMI', 'Age', 'Daily_Calorie_Intake',
            'Daily_Protein_Intake_(g)', 'Daily_Fat_Intake_(g)', 'Daily_Carbohydrate_Intake_(g)',
            'Daily_Fiber_Intake_(g)', 'Daily_Water_Intake_(liters)', 'Food_Diversity_Score',
            'Hemoglobin_(g/dL)', 'Serum_Ferritin_(ng/mL)', 'Vitamin_D_25-OH_(ng/mL)',
            'Serum_B12_(pg/mL)', 'Serum_Folate_(ng/mL)', 'C-Reactive_Protein_(mg/dL)',
            'Serum_Albumin_(g/dL)', 'Total_Cholesterol_(mg/dL)', 'Fasting_Glucose_(mg/dL)',
            'Alcohol_Consumption_(g/day)', 'Medication_Interaction_Score'
        ])
]

# Fill missing values in numeric columns with 0
df[numeric_cols] = df[numeric_cols].fillna(0)

# Save cleaned dataset
df.to_csv(r'C:\My stuff\Coding\ML project\KiranveerSingh_Project\Dataset\Final_Nutrition_Dataset.csv', index=False)
print("Cleaned dataset saved as Final_Nutrition_Dataset.csv")
```
